# Python String Algorithms & Manipulation Exercises

This notebook contains Python implementations of common string manipulation algorithms, substring searches, palindromic checks, and combinatorial generation techniques—highly useful for coding interviews and technical preparation.

```python
s = "ABC"
# Iterate through each character in the string by its index and print it
for i in range(len(s)):
  print(s[i])
```

```python
s = "ABC"
# Generate and print all possible 2-character permutations (with repetition) from the string
for i in range(len(s)):
  for j in range(len(s)):
    print(s[i]+s[j])
```

```python
s = "ABC"
# Generate and print all unique pairs of characters (combinations without repetition) where order doesn't matter
for i in range(len(s)):
  for j in range(i+1, len(s)):
    print(s[i]+s[j])
```

```python
s = "ABC"
# Generate and print all substrings using string slicing with an exclusive upper bound
for i in range(len(s)):
  for j in range(i+1, len(s)+1):
    print(s[i:j])
```

```python
s = "ABC"
print("All Substrings:")
# Generate and print all substrings using an inclusive index upper bound (j+1)
for i in range(len(s)):
  for j in range(i, len(s)):
    print(s[i:j+1])
```

```python
s = "ABC"
c = 0
# Count the total number of substrings that can be formed from the string
for i in range(len(s)):
  for j in range(i+1, len(s)+1):
    c += 1
print(c)
```

```python
s = "ksjdfhsuiywoqweupqr"
c = 0
# Count and print the number of vowels present in the given string
for i in s:
  if i in "aeiou":
    c += 1
print(c)
```

```python
s = "madam"
n = len(s)
is_palindrome = True
# Check if the string is a palindrome by comparing characters from both ends moving inward
for i in range(len(s)):
  if s[i] != s[n-i-1]:
    is_palindrome = False
    break
print(is_palindrome)
```

```python
s = "aba"
# Find and print all substrings where the first and last characters are identical
for i in range(len(s)):
  for j in range(i+1, len(s)+1):
    sub = s[i:j]
    if sub[0] == sub[-1]:
      print(sub)
```

```python
s = "abccccbd"
longest = ""
# Find the longest palindromic substring by checking every possible substring
for i in range(len(s)):
  for j in range(i, len(s)):
    sub = s[i:j+1]
    # Check if the substring is equal to its reverse and is longer than the current longest
    if sub == sub[::-1] and len(sub) > len(longest):
      longest = sub
print(longest)
```

```python
def lengthOfLongestSubstring(s):
  max_len = 0
  longest = ""
  # Helper function to find the longest substring with unique (non-repeating) characters
  for i in range(len(s)):
    for j in range(i, len(s)):
      sub = s[i:j+1]
      # If the number of unique elements (set) equals the length of the substring, it has no duplicates
      if len(set(sub)) == len(sub) and len(sub) > max_len:
        longest = sub
        max_len = len(longest)
  return longest

# Test the function with a sample string
lengthOfLongestSubstring("abcabcbb")
```
